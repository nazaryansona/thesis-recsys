```python
import os
base = "../data/MOOCCube"
for root, dirs, files in os.walk(base):
    for name in sorted(files):
        if name.startswith("."):
            continue
        p = os.path.join(root, name)
        print(os.path.relpath(p, base), round(os.path.getsize(p) / 1e6, 1), "MB")
```

```python
import os
base = "../data/MOOCCube"
for folder in ["entities", "relations", "additional_information"]:
    print("==", folder)
    for name in sorted(os.listdir(os.path.join(base, folder))):
        p = os.path.join(base, folder, name)
        print(name, round(os.path.getsize(p) / 1e6, 1), "MB")
```

```python
def peek(rel_path, n=3, width=400):
    with open(os.path.join(base, rel_path), encoding="utf-8") as f:
        for _ in range(n):
            print(f.readline()[:width])
            print()

peek("entities/concept.json")
peek("relations/user-course.json")
peek("relations/course-concept.json")
```

```python
import pandas as pd
base = "../data/MOOCCube"

uc = pd.read_csv(f"{base}/relations/user-course.json", sep="\t", header=None, names=["user", "course"])
cc = pd.read_csv(f"{base}/relations/course-concept.json", sep="\t", header=None, names=["course", "concept"])

print("user-course rows:", len(uc), "| users:", uc["user"].nunique(), "| courses:", uc["course"].nunique())
print("course-concept rows:", len(cc), "| courses:", cc["course"].nunique(), "| concepts:", cc["concept"].nunique())
print(uc.groupby("user").size().describe())
```

```python
concepts = pd.read_json(f"{base}/entities/concept.json", lines=True)
real = concepts[~concepts["id"].str.startswith("K_T_")]
print(len(concepts), len(real))
print("share with English name:", real["en"].notna().mean())
```

```python
pop = uc.groupby("course").size().sort_values(ascending=False)
print("top 10 courses, share of all enrolments:", round(pop.head(10).sum() / len(uc), 3))
print("users with 3 or more courses:", (uc.groupby("user").size() >= 3).sum())
print("enrolled courses with no concepts:", len(set(uc["course"]) - set(cc["course"])))
print("real concepts with a non-empty English name:", round((real["en"].str.strip() != "").mean(), 3))
print(cc.groupby("course").size().describe())
```

```python
courses_ok = set(cc["course"])
df = uc[uc["course"].isin(courses_ok)].drop_duplicates()
n = df.groupby("user")["course"].transform("size")
df = df[n >= 3]
print("rows:", len(df), "| users:", df["user"].nunique(), "| courses:", df["course"].nunique())
os.makedirs("../data/processed", exist_ok=True)
df.to_csv("../data/processed/interactions.csv", index=False)
```
